```python
import requests
from bs4 import BeautifulSoup
import pandas

url='https://books.toscrape.com/'
response=requests.get(url)
soup=BeautifulSoup(response.text,'html.parser')
books=soup.find_all('article',class_='product_pod')

# Initialize data list before using it
data = []

for book in books:
    title = book.find('h3').a['title']
    price_str = book.find('p', class_='price_color').text

    # Extract rating. The rating is usually in the class attribute of a <p> tag.
    rating_tag = book.find('p', class_=lambda x: x and x.startswith('star-rating'))
    rating = rating_tag['class'][1] if rating_tag else 'No Rating'

    price_value = float(price_str.replace('Â£', ''))

    data.append({'Title': title, 'Price': price_value, 'Rating': rating})


df = pandas.DataFrame(data)

# Filter for 5-star and 4-star rated books
top_rated_books = df[df['Rating'].isin(['Five', 'Four'])]

print("Top Rated Books (5-star and 4-star):")
for index, row in top_rated_books.iterrows():
    print(f"- {row['Title']} (Rating: {row['Rating']})")
```

```python
import sqlite3
import pandas as pd
import random

def fetch_data_database():
  conn = None # Initialize conn to None
  try:
    conn = sqlite3.connect('example.db')
    cursor = conn.cursor()

    cursor.execute('''
    CREATE TABLE IF NOT EXISTS USERS (
    ID INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER
    )
    ''')
    conn.commit()

    # Clear existing data to ensure a fresh set of 20 users and prevent IntegrityError
    cursor.execute('DELETE FROM USERS')
    conn.commit()

    # Insert 20 sample values for ID, name, and age
    for i in range(1, 21):
        user_id = i
        user_name = f'User_{i}'
        user_age = 20 + i # Just an example age
        cursor.execute("INSERT INTO USERS (ID, name, age) VALUES (?, ?, ?)", (user_id, user_name, user_age))

    conn.commit()
    print("20 users inserted successfully into the USERS table.")

    # Now, fetch the data and create DataFrame within the function's scope
    cursor.execute("SELECT * FROM USERS")
    rows = cursor.fetchall()

    df = pd.DataFrame(rows, columns=['ID', 'Name', 'Age'])


    df['Marks'] = [random.randint(50, 100) for _ in range(len(df))]
    print("\nDataFrame with Marks column:")
    print(df)


    df_sorted_by_age = df.sort_values(by='Age', ascending=True)
    print("\nDataFrame sorted by Age (ascending):")
    print(df_sorted_by_age)


    mean_marks = df['Marks'].mean()
    print(f"\nMean Marks: {mean_marks:.2f}")


    candidates_above_mean = df[df['Marks'] > mean_marks]
    print("\nCandidates who scored more than the mean marks:")
    if not candidates_above_mean.empty:
        print(candidates_above_mean[['Name', 'Marks']])
    else:
        print("No candidates scored above the mean.")
    return df

  except sqlite3.OperationalError as e:
    print(f"Database error: {e}. You might need to restart the Colab runtime to release the lock.")
    return None
  except Exception as e:
    print(f"An unexpected error occurred: {e}")
    return None
  finally:
    if conn:
      conn.close()

# Call the function to create table and insert data and assign the returned DataFrame to df
df = fetch_data_database()
```

#### Removing `\r\n` characters from the 'phone' column using string slicing

To clean the 'phone' column, we will remove any trailing `\r\n` characters using string slicing. This method is effective when the characters to be removed are consistently at the end of the string. We first convert the column to string type to safely handle `NaN` values and other data types before applying string operations.

```python

display(df['phone'].sample(5))

df['phone'] = df['phone'].astype(str).apply(
    lambda x: x[:-2] if x.endswith('\r\n') else x
)

#after slicing
display(df['phone'].sample(5))

display(df['phone'].head())
```

#### Splitting the 'phone' column into two new columns

If a 'phone' entry contains multiple numbers, we'll split them into 'phone_1' and 'phone_2' columns. We'll attempt to split by comma (`,`) or newline ( `\n` ) as these are common delimiters for multiple values within a single field. Entries with only one number will have `NaN` in the second phone column.

```python
# First, inspect a few entries that might contain multiple phone numbers
# To do this, let's look for entries containing a common delimiter like ',' or '\n'
# We'll sample 10 such entries if they exist.

print("Sample of 'phone' entries that might contain multiple numbers:")
multi_phone_entries = df['phone'].astype(str).loc[
    df['phone'].astype(str).str.contains(',', na=False) | df['phone'].astype(str).str.contains('\n', na=False)
]

if not multi_phone_entries.empty:
    display(multi_phone_entries.sample(min(10, len(multi_phone_entries))))
else:
    print("No entries found with comma or newline separators.")

# Split the 'phone' column by comma or newline. This will create a list of strings.
# Then, expand this list into new columns.
# We will use str.split with expand=True, which creates new columns for each split part.
# If there's only one number, the subsequent columns will be filled with NaN.

# Prioritize splitting by comma, then by newline if comma is not present.
# This function will handle both cases.
def split_phone_numbers(phone_str):
    if pd.isna(phone_str) or not isinstance(phone_str, str):
        return [None, None]
    if ',' in phone_str:
        parts = phone_str.split(',', 1)
    elif '\n' in phone_str:
        parts = phone_str.split('\n', 1)
    else:
        return [phone_str.strip(), None]

    # Ensure we return exactly two elements, padding with None if necessary
    return [p.strip() if p else None for p in parts] + [None] * (2 - len(parts))

# Apply the splitting function
split_phones = df['phone'].apply(split_phone_numbers).apply(pd.Series)
split_phones.columns = ['phone_1', 'phone_2']

# Add the new columns to the original DataFrame
df = pd.concat([df, split_phones], axis=1)

# Display the DataFrame with the new 'phone_1' and 'phone_2' columns
print("\nDataFrame with 'phone_1' and 'phone_2' columns:")
display(df[['phone', 'phone_1', 'phone_2']].head())
```

```python
df['phone'] = df['phone'].astype(str)
df[["phone1", "phone2"]] = df["phone"].str.split(r"\r\n", expand=True)
df["phone2"] = df["phone_2"].fillna(df["phone_1"])
display(df.head(2))
```
